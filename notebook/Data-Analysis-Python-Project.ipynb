```python
import pandas as pd

df = pd.read_csv(r"/Users/britneyspiercegnanadhas/Downloads/raw_dataset.csv", sep=";")

print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
df["ball-bearing"] = pd.to_numeric(df["ball-bearing"], errors="coerce")
df["humidity"] = pd.to_numeric(df["humidity"], errors="coerce")
df["vibration"] = pd.to_numeric(df["vibration"], errors="coerce")
```

```python
print(df.dtypes)
```

```python
print("\nMissing values after conversion:")
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
print("\nFinal shape:", df.shape)
print("Final duplicates:", df.duplicated().sum())
```

```python
print("\nDataset summary:")
print(df.describe())
```

```python
print("\nMissing values conversion:")
print(df.isnull().sum())
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_dataset.csv")

print(df.columns.tolist())
print(df.head())
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df["ball-bearing"])

plt.title("Ball-Bearing Trend")
plt.xlabel("Observation")
plt.ylabel("Ball-Bearing")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df["humidity"])

plt.title("Humidity Trend")
plt.xlabel("Observation")
plt.ylabel("Humidity")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df["vibration"])

plt.title("Vibration Trend")
plt.xlabel("Observation")
plt.ylabel("Vibration")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["ball-bearing"],
    df["vibration"],
    alpha=0.3
)

plt.title("Ball-Bearing vs Vibration")
plt.xlabel("Ball-Bearing")
plt.ylabel("Vibration")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["humidity"],
    df["vibration"],
    alpha=0.3
)

plt.title("Humidity vs Vibration")
plt.xlabel("Humidity")
plt.ylabel("Vibration")
plt.grid(True)

plt.show()
```
